```python
# uma função é um método
# para executar algo muitas vezes

def boas_vindas():
  print(' ----Bem Vindo ao sistema----')
  nome = input('Me informe seu nome: ')
  print(f'Olá, {nome} o que vamos fazer hoje?')

def boas_vindas_usuario(usuario):
  print(f"Olá {usuario}!")
```

```python
usuario = input("Me informe seu nome: ")
boas_vindas_usuario(usuario)
```

```python
# Vamos criar um sistema que leia um algoritmo
# No Brasil a temperatura é em °C
# Nos EUA é em Fahrenheit (°F)
# Mas o certo K - Kelvin

def celcius_f(temperatura):
  # f = (valor em °C *1.8 ) + 32
  f = (temperatura * 1.8) + 32
  # dentro de uma função usamos o comando RETURN
  return f

def celcius_k(temperatura):
  k = temperatura + 273
  return k
```

```python
temperatura = float(input("informe a temperatura: "))
print(celcius_f(temperatura))
print(celcius_k(temperatura))
```
